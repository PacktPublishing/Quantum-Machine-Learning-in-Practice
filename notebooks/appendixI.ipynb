```python
import numpy as np

def is_normalized(v, tol=1e-8):
    return np.allclose(np.vdot(v, v), 1.0, atol=tol)

def is_hermitian(A, tol=1e-8):
    return np.allclose(A, A.conj().T, atol=tol)

def is_unitary(U, tol=1e-8):
    n = U.shape[0]
    return np.allclose(U.conj().T @ U, np.eye(n), atol=tol)

def symmetrize(K):
    return 0.5 * (K + K.conj().T)

def min_psd_eigenvalue(K):
    return np.linalg.eigvalsh(symmetrize(K)).min()

def condition_number(A):
    s = np.linalg.svd(A, compute_uv=False)
    return s[0] / s[-1]
```

```python
# Runtime checks for the Appendix I listing.
ket0 = np.array([1.0 + 0.0j, 0.0 + 0.0j])
ket_plus = np.array([1.0, 1.0], dtype=complex) / np.sqrt(2)

H = np.array([[1.0, 1.0], [1.0, -1.0]], dtype=complex) / np.sqrt(2)
Z = np.array([[1.0, 0.0], [0.0, -1.0]], dtype=complex)
K = np.array([[1.0, 0.25], [0.25, 1.0]], dtype=complex)
A = np.array([[3.0, 0.0], [0.0, 1.0]])

assert is_normalized(ket0)
assert is_normalized(ket_plus)
assert is_hermitian(Z)
assert is_unitary(H)
assert min_psd_eigenvalue(K) >= -1e-8
assert np.isclose(condition_number(A), 3.0)

print("Appendix I checks passed.")
```
