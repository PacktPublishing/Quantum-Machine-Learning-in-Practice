```python
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

X, y = make_classification(
    n_samples=1200,
    n_features=12,
    n_informative=8,
    n_redundant=2,
    class_sep=1.1,
    random_state=42,
)

models = {
    "logistic_regression": make_pipeline(
        StandardScaler(),
        LogisticRegression(max_iter=2000)
    ),
    "rbf_svm": make_pipeline(
        StandardScaler(),
        SVC(kernel="rbf")
    ),
    "random_forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        random_state=42,
        n_jobs=-1,
    ),
}

scoring = {
    "accuracy": "accuracy",
    "f1": "f1",
    "roc_auc": "roc_auc",
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

results = {}
for name, model in models.items():
    scores = cross_validate(model, X, y, scoring=scoring, cv=cv, n_jobs=-1)
    results[name] = {metric: scores[f"test_{metric}"].mean() for metric in scoring}

for name, metrics in results.items():
    print(name, metrics)
```

```python
from dataclasses import dataclass
from typing import Optional

@dataclass
class ExperimentCard:
    model_name: str
    validation_metric: float
    test_metric: float
    wall_clock_seconds: float
    parameter_count: int
    simulator_type: Optional[str] = None
    qubits: Optional[int] = None
    circuit_depth: Optional[int] = None
    shots: Optional[int] = None
    notes: str = ""

card = ExperimentCard(
    model_name="quantum_kernel_svm",
    validation_metric=0.842,
    test_metric=0.836,
    wall_clock_seconds=91.4,
    parameter_count=0,
    simulator_type="statevector",
    qubits=6,
    circuit_depth=18,
    shots=None,
    notes="Compared against tuned RBF SVM and logistic regression baselines."
)

print(card)
```
